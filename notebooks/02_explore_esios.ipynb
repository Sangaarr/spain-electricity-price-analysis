```python
import os
from dotenv import load_dotenv
load_dotenv("../.env")
token=os.getenv("ESIOS_TOKEN")
print(len(token))
```

```python
import requests

headers = {
    "Accept": "application/json; application/vnd.esios-api-v1+json",
    "Content-Type": "application/json",
    "x-api-key": token,
}

url = "https://api.esios.ree.es/indicators/1293?start_date=2024-01-01T00:00:00&end_date=2024-01-01T23:59:59"
response = requests.get(url, headers=headers)
print(response.status_code)
```

```python
data = response.json()
```
